```python
import sys; sys.path.insert(0, "..")
import pandas as pd
import matplotlib.pyplot as plt
from data.fetcher import fetch_yfinance
from src.indicators import rsi, macd, bollinger_bands, atr, adx, supertrend
plt.style.use("seaborn-v0_8-darkgrid")
plt.rcParams["figure.figsize"] = (14, 5)
```

```python
# Fetch AAPL daily data
df = fetch_yfinance("AAPL", start="2022-01-01", end="2024-01-01")
df.tail()
```

```python
# RSI
rsi_14 = rsi(df["Close"], period=14)
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)
ax1.plot(df["Close"], label="AAPL Close")
ax2.plot(rsi_14, color="purple")
ax2.axhline(70, linestyle="--", color="red", alpha=0.5)
ax2.axhline(30, linestyle="--", color="green", alpha=0.5)
ax2.set_ylabel("RSI(14)")
plt.tight_layout(); plt.show()
```

```python
# MACD
macd_df = macd(df["Close"])
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)
ax1.plot(df["Close"])
ax2.plot(macd_df["MACD"], label="MACD")
ax2.plot(macd_df["Signal"], label="Signal")
ax2.bar(macd_df.index, macd_df["Histogram"],
        color=["green" if x>0 else "red" for x in macd_df["Histogram"]], alpha=0.4)
ax2.legend(); plt.tight_layout(); plt.show()
```

```python
# Bollinger Bands
bb = bollinger_bands(df["Close"])
plt.figure()
plt.plot(df["Close"], label="Close", color="black", lw=1)
plt.plot(bb["BB_Upper"], "--", color="red", label="Upper")
plt.plot(bb["BB_Middle"], "--", color="blue", label="Middle")
plt.plot(bb["BB_Lower"], "--", color="green", label="Lower")
plt.fill_between(bb.index, bb["BB_Lower"], bb["BB_Upper"], alpha=0.08, color="blue")
plt.legend(); plt.title("Bollinger Bands"); plt.show()
```

```python
# ADX – trend strength
adx_df = adx(df["High"], df["Low"], df["Close"])
plt.figure()
plt.plot(adx_df["ADX"], label="ADX", color="navy")
plt.axhline(25, linestyle="--", color="red", alpha=0.4, label="Threshold 25")
plt.legend(); plt.title("ADX Trend Strength"); plt.show()
```
